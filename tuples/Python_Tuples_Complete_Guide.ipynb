## 1. What is a Tuple?

A **Tuple** is a collection used to store multiple values in a single variable. Tuples are commonly created using parentheses `()`.

```python
fruits = ("apple", "banana", "mango")

print(fruits)
```

## 2. Creating a Tuple

A tuple can contain multiple values separated by commas.

```python
numbers = (10, 20, 30, 40)

print(numbers)
```

## 3. Single-Item Tuple

A single-item tuple needs a comma. Without the comma, Python treats it as a normal value.

```python
single_tuple = ("apple",)

not_a_tuple = ("apple")

print(single_tuple)
print(type(single_tuple))
print(type(not_a_tuple))
```

## 4. Different Data Types in a Tuple

A tuple can contain different data types.

```python
student = ("Shabbir", 15, 85.5, True)

print(student)
```

## 5. Tuples are Ordered

Tuple items keep their order and each item has an index.

```python
fruits = ("apple", "banana", "mango")

print(fruits)
```

## 6. Tuples are Immutable

**Immutable** means that tuple items cannot be changed after the tuple is created.

```python
fruits = ("apple", "banana", "mango")

# The following line would raise a TypeError:
# fruits[1] = "orange"

print(fruits)
```

## 7. Positive Indexing

Indexes start from `0`.

```python
fruits = ("apple", "banana", "mango", "orange")

print(fruits[0])
print(fruits[2])
```

## 8. Negative Indexing

Negative indexing starts from the end. `-1` represents the last item.

```python
fruits = ("apple", "banana", "mango", "orange")

print(fruits[-1])
print(fruits[-2])
```

## 9. Tuple Slicing

Slicing extracts a portion of a tuple.

Syntax: `tuple[start:stop:step]`

The `stop` index is not included.

```python
numbers = (10, 20, 30, 40, 50)

print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
print(numbers[::2])
```

## 10. Tuple Length

`len()` returns the number of items in a tuple.

```python
fruits = ("apple", "banana", "mango")

print(len(fruits))
```

## 11. Checking Membership

Use `in` and `not in` to check whether a value exists in a tuple.

```python
fruits = ("apple", "banana", "mango")

print("banana" in fruits)
print("orange" not in fruits)
```

## 12. `count()`

Returns how many times a value occurs in a tuple.

```python
numbers = (10, 20, 20, 30, 20)

print(numbers.count(20))
```

## 13. `index()`

Returns the index of the first matching value.

```python
fruits = ("apple", "banana", "mango")

print(fruits.index("mango"))
```

## 14. Tuple Packing

Putting multiple values into a tuple is called **packing**.

```python
student = "Shabbir", 15, "Python"

print(student)
print(type(student))
```

## 15. Tuple Unpacking

Taking values from a tuple and assigning them to separate variables is called **unpacking**.

```python
student = ("Shabbir", 15, "Python")

name, age, subject = student

print(name)
print(age)
print(subject)
```

## 16. Unpacking with `*`

The `*` operator can collect multiple remaining values into a list.

```python
numbers = (10, 20, 30, 40, 50)

first, *middle, last = numbers

print("First:", first)
print("Middle:", middle)
print("Last:", last)
```

## 17. Loop Through a Tuple

A `for` loop can visit each item one by one.

```python
fruits = ("apple", "banana", "mango")

for fruit in fruits:
    print(fruit)
```

## 18. Loop Using Indexes

`range()` and `len()` can be used when you need the index.

```python
fruits = ("apple", "banana", "mango")

for i in range(len(fruits)):
    print(i, fruits[i])
```

## 19. Nested Tuples

A tuple can contain other tuples. This is called a **nested tuple**.

```python
students = (
    ("Shabbir", 85),
    ("Ali", 90),
    ("Ahmed", 78)
)

print(students)
print(students[0])
print(students[0][0])
```

## 20. Concatenating Tuples

The `+` operator can combine two tuples into a new tuple.

```python
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)

combined = tuple1 + tuple2

print(combined)
```

## 21. Repeating Tuples

The `*` operator can repeat the items of a tuple.

```python
numbers = (1, 2, 3)

result = numbers * 2

print(result)
```

## 22. Useful Built-in Functions

These Python built-in functions are commonly used with numeric tuples.

```python
numbers = (10, 20, 30, 40)

print("Length:", len(numbers))
print("Maximum:", max(numbers))
print("Minimum:", min(numbers))
print("Sum:", sum(numbers))
print("Sorted:", sorted(numbers))
```

## 23. Converting List to Tuple

The `tuple()` function can convert another iterable, such as a list, into a tuple.

```python
fruits_list = ["apple", "banana", "mango"]

fruits_tuple = tuple(fruits_list)

print(fruits_tuple)
print(type(fruits_tuple))
```

## 24. Converting Tuple to List

The `list()` function can convert a tuple into a list. This is useful if you need to modify the data.

```python
fruits_tuple = ("apple", "banana", "mango")

fruits_list = list(fruits_tuple)

fruits_list[1] = "orange"

print(fruits_list)
```

## 🧠 Tuple Methods — Quick Revision

Tuples have only **two built-in methods**:

| Method | Use |
|---|---|
| `count()` | Counts how many times a value occurs |
| `index()` | Finds the index of the first matching value |

### Useful built-in functions

- `len()` — number of items
- `max()` — largest value
- `min()` — smallest value
- `sum()` — total of numeric values
- `sorted()` — returns a sorted list
- `tuple()` — converts an iterable to a tuple
- `list()` — converts a tuple to a list


## 🆚 List vs Tuple

| Feature | List | Tuple |
|---|---|---|
| Brackets | `[]` | `()` |
| Mutable | ✅ Yes | ❌ No |
| Ordered | ✅ Yes | ✅ Yes |
| Indexing | ✅ Yes | ✅ Yes |
| Slicing | ✅ Yes | ✅ Yes |
| `append()` | ✅ Yes | ❌ No |
| `extend()` | ✅ Yes | ❌ No |
| `remove()` | ✅ Yes | ❌ No |
| `count()` | ✅ Yes | ✅ Yes |
| `index()` | ✅ Yes | ✅ Yes |

**Main difference:** Lists can be changed, while tuples cannot be changed after creation.
