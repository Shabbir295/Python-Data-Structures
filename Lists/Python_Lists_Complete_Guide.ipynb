## 1. What is a List?

A **List** is a collection used to store multiple values in a single variable. Lists are created using square brackets `[]`.

```python
fruits = ["apple", "banana", "mango"]

print(fruits)
```

## 2. Different Data Types in a List

A list can contain values of different data types.

```python
student = ["Shabbir", 15, 85.5, True]

print(student)
```

## 3. Lists are Ordered

List items keep their order. Each item has a position called an **index**.

```python
fruits = ["apple", "banana", "mango"]

print(fruits)
```

## 4. Lists are Mutable

Mutable means that list items can be changed after the list is created.

```python
fruits = ["apple", "banana", "mango"]

fruits[1] = "orange"

print(fruits)
```

## 5. Positive Indexing

Indexes start from `0`.

```python
fruits = ["apple", "banana", "mango", "orange"]

print(fruits[0])
print(fruits[2])
```

## 6. Negative Indexing

Negative indexing starts from the end. `-1` represents the last item.

```python
fruits = ["apple", "banana", "mango", "orange"]

print(fruits[-1])
print(fruits[-2])
```

## 7. List Slicing

Slicing extracts a portion of a list.

Syntax: `list[start:stop:step]`

The `stop` index is not included.

```python
numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
print(numbers[::2])
```

## 8. Updating List Items

You can change one or multiple items using indexes or slices.

```python
numbers = [10, 20, 30, 40]

numbers[1] = 200
numbers[2:4] = [300, 400]

print(numbers)
```

## 9. Checking Membership

Use `in` and `not in` to check whether a value exists in a list.

```python
fruits = ["apple", "banana", "mango"]

print("banana" in fruits)
print("orange" not in fruits)
```

## 10. `append()`

Adds **one item** to the end of a list.

```python
fruits = ["apple", "banana"]

fruits.append("mango")

print(fruits)
```

## 11. `extend()`

Adds multiple items from another iterable to the end of a list.

```python
fruits = ["apple", "banana"]

fruits.extend(["mango", "orange"])

print(fruits)
```

### `append()` vs `extend()`

```python
fruits = ["apple", "banana"]

fruits.append(["mango", "orange"])
print("append:", fruits)

fruits = ["apple", "banana"]

fruits.extend(["mango", "orange"])
print("extend:", fruits)
```

## 12. `insert()`

Adds an item at a specific index.

```python
fruits = ["apple", "banana", "mango"]

fruits.insert(1, "orange")

print(fruits)
```

## 13. `remove()`

Removes the first occurrence of a specific value.

```python
fruits = ["apple", "banana", "mango", "banana"]

fruits.remove("banana")

print(fruits)
```

## 14. `pop()`

Removes and returns an item by index. If no index is given, it removes the last item.

```python
fruits = ["apple", "banana", "mango"]

removed_item = fruits.pop(1)

print("Removed:", removed_item)
print("List:", fruits)

fruits.pop()
print("After removing last item:", fruits)
```

## 15. `clear()`

Removes all items from a list.

```python
fruits = ["apple", "banana", "mango"]

fruits.clear()

print(fruits)
```

## 16. `index()`

Returns the index of the first matching value.

```python
fruits = ["apple", "banana", "mango"]

print(fruits.index("mango"))
```

## 17. `count()`

Returns how many times a value occurs in a list.

```python
numbers = [10, 20, 20, 30, 20]

print(numbers.count(20))
```

## 18. `sort()`

Sorts a list in ascending order. Use `reverse=True` for descending order.

```python
numbers = [40, 10, 30, 20]

numbers.sort()
print("Ascending:", numbers)

numbers.sort(reverse=True)
print("Descending:", numbers)
```

## 19. `reverse()`

Reverses the current order of items.

```python
numbers = [10, 20, 30, 40]

numbers.reverse()

print(numbers)
```

## 20. `copy()`

Creates a shallow copy of a list.

```python
numbers = [10, 20, 30]

new_numbers = numbers.copy()

print("Original:", numbers)
print("Copy:", new_numbers)
```

## 21. `len()`

Returns the number of items in a list.

```python
fruits = ["apple", "banana", "mango"]

print(len(fruits))
```

## 22. Loop Through a List

A `for` loop can visit each item one by one.

```python
fruits = ["apple", "banana", "mango"]

for fruit in fruits:
    print(fruit)
```

## 23. Loop Using Indexes

`range()` and `len()` can be used when you need the index.

```python
fruits = ["apple", "banana", "mango"]

for i in range(len(fruits)):
    print(i, fruits[i])
```

## 24. Nested Lists

A list can contain other lists. This is called a **nested list**.

```python
students = [
    ["Shabbir", 85],
    ["Ali", 90],
    ["Ahmed", 78]
]

print(students)
print(students[0])
print(students[0][0])
```

## 25. Useful Built-in Functions

These are Python built-in functions commonly used with numeric lists.

```python
numbers = [10, 20, 30, 40]

print("Length:", len(numbers))
print("Maximum:", max(numbers))
print("Minimum:", min(numbers))
print("Sum:", sum(numbers))
print("Sorted:", sorted(numbers))
```

- `max()` — largest value
- `min()` — smallest value
