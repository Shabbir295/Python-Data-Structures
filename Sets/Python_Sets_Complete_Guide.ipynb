# Python Sets — Complete Guide

A complete beginner-friendly guide to Python Sets, from basics to practice.


## 1. What is a Set?

A **set** is a collection in Python that stores **unique values**. Sets are unordered, so items do not have a fixed position or index.


```python
numbers = {10, 20, 30, 40}
print(numbers)
```

## 2. Creating a Set


```python
fruits = {"apple", "banana", "mango"}
print(fruits)
```

## 3. Duplicate Values

Sets automatically remove duplicate values.

```python
numbers = {1, 2, 2, 3, 3, 4}
print(numbers)
```

## 4. Empty Set

Use `set()` to create an empty set. `{}` creates an empty dictionary.

```python
empty_set = set()
print(empty_set)
```

## 5. Different Data Types

A set can contain values of different data types.

```python
data = {10, "Python", 3.14, True}
print(data)
```

## 6. Length of a Set

Use `len()` to find the number of unique items.

```python
students = {"Ali", "Ahmed", "Shabbir", "Zain"}
print(len(students))
```

## 7. Add an Item

Use `add()` to add one item.

```python
fruits = {"apple", "banana"}
fruits.add("mango")
print(fruits)
```

## 8. Update a Set

Use `update()` to add multiple items.

```python
numbers = {1, 2, 3}
numbers.update({4, 5, 6})
print(numbers)
```

## 9. Remove an Item

`remove()` deletes an item. It raises an error if the item does not exist.

```python
fruits = {"apple", "banana", "mango"}
fruits.remove("banana")
print(fruits)
```

## 10. Discard an Item

`discard()` removes an item without raising an error if it is missing.

```python
fruits = {"apple", "banana", "mango"}
fruits.discard("orange")
print(fruits)
```

## 11. Pop an Item

`pop()` removes and returns an arbitrary item because sets are unordered.

```python
numbers = {10, 20, 30}
removed = numbers.pop()
print("Removed:", removed)
print("Set:", numbers)
```

## 12. Clear a Set

`clear()` removes all items.

```python
numbers = {1, 2, 3}
numbers.clear()
print(numbers)
```

## 13. Membership Testing

Use `in` and `not in` to check whether an item exists.

```python
fruits = {"apple", "banana", "mango"}
print("apple" in fruits)
print("orange" not in fruits)
```

## 14. Union

Union combines all unique items from two sets.

```python
a = {1, 2, 3}
b = {3, 4, 5}
print(a.union(b))
print(a | b)
```

## 15. Intersection

Intersection returns items common to both sets.

```python
a = {1, 2, 3}
b = {3, 4, 5}
print(a.intersection(b))
print(a & b)
```

## 16. Difference

Difference returns items that are in the first set but not in the second.

```python
a = {1, 2, 3}
b = {3, 4, 5}
print(a.difference(b))
print(a - b)
```

## 17. Symmetric Difference

Returns items that are in either set, but not in both.

```python
a = {1, 2, 3}
b = {3, 4, 5}
print(a.symmetric_difference(b))
print(a ^ b)
```

## 18. Subset

A set is a subset when all its items are present in another set.

```python
a = {1, 2}
b = {1, 2, 3, 4}
print(a.issubset(b))
```

## 19. Superset

A set is a superset when it contains all items of another set.

```python
a = {1, 2, 3, 4}
b = {1, 2}
print(a.issuperset(b))
```

## 20. Disjoint Sets

Two sets are disjoint when they have no common items.

```python
a = {1, 2, 3}
b = {4, 5, 6}
print(a.isdisjoint(b))
```

## 21. Frozenset

A `frozenset` is an immutable version of a set. It cannot be changed after creation.

```python
numbers = frozenset({1, 2, 3})
print(numbers)
```

## 22. Important Set Rules

- Sets store unique values.
- Sets are unordered.
- Sets do not support indexing or slicing.
- Sets are mutable.
- Set elements must be hashable.


## 23. Real-World Example

Finding common students between two courses:

```python
python_students = {"Ali", "Ahmed", "Shabbir", "Zain"}
ai_students = {"Shabbir", "Zain", "Hamza"}

common_students = python_students.intersection(ai_students)
print("Common students:", common_students)
```

## 25. Mini Challenge — Student Skills

Create two sets containing skills of two students. Find:

- Skills both students have
- Skills only Student A has
- Skills only Student B has
- All unique skills


```python
student_a = {"Python", "SQL", "Git", "HTML"}
student_b = {"Python", "Git", "JavaScript", "CSS"}

print("Common skills:", student_a & student_b)
print("Only Student A:", student_a - student_b)
print("Only Student B:", student_b - student_a)
print("All skills:", student_a | student_b)
```
