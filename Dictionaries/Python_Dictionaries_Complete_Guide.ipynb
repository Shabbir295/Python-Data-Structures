## 1. What is a Dictionary?

A **Dictionary** is a collection that stores data in **key-value pairs**. Dictionaries are created using curly brackets `{}`.

```python
student = {
    "name": "Shabbir",
    "age": 15,
    "course": "Python"
}

print(student)
```

## 2. Key-Value Pairs

Each item in a dictionary has a **key** and a corresponding **value**.

Example: `"name"` is the key and `"Shabbir"` is its value.

```python
student = {
    "name": "Shabbir",
    "age": 15,
    "course": "Python"
}

print(student["name"])
print(student["age"])
```

## 3. Different Data Types

Dictionary values can have different data types.

```python
student = {
    "name": "Shabbir",
    "age": 15,
    "marks": 85.5,
    "passed": True
}

print(student)
```

## 4. Dictionary with a List as a Value

A dictionary value can also be a list or another collection.

```python
student = {
    "name": "Shabbir",
    "subjects": ["Python", "Math", "English"]
}

print(student["subjects"])
```

## 5. Accessing Values with `[]`

Use the key inside square brackets to access its value.

```python
student = {
    "name": "Shabbir",
    "age": 15
}

print(student["name"])
print(student["age"])
```

## 6. Accessing Values with `get()`

`get()` returns the value for a key. It can also provide a default value if the key does not exist.

```python
student = {
    "name": "Shabbir",
    "age": 15
}

print(student.get("name"))
print(student.get("email"))
print(student.get("email", "Not Found"))
```

## 7. Adding a New Item

Assign a value to a new key to add an item.

```python
student = {
    "name": "Shabbir",
    "age": 15
}

student["course"] = "Python"

print(student)
```

## 8. Updating a Value

Assign a new value to an existing key to update it.

```python
student = {
    "name": "Shabbir",
    "age": 15
}

student["age"] = 16

print(student)
```

## 9. `update()`

`update()` can add new key-value pairs or update existing ones.

```python
student = {
    "name": "Shabbir",
    "age": 15
}

student.update({
    "age": 16,
    "course": "Python"
})

print(student)
```

## 10. Removing Items with `pop()`

`pop()` removes a specific key and returns its value.

```python
student = {
    "name": "Shabbir",
    "age": 15,
    "course": "Python"
}

removed = student.pop("age")

print("Removed:", removed)
print("Dictionary:", student)
```

## 11. `popitem()`

`popitem()` removes and returns the last inserted key-value pair.

```python
student = {
    "name": "Shabbir",
    "age": 15,
    "course": "Python"
}

removed = student.popitem()

print("Removed:", removed)
print("Dictionary:", student)
```

## 12. `del`

`del` can remove a specific key-value pair.

```python
student = {
    "name": "Shabbir",
    "age": 15,
    "course": "Python"
}

del student["age"]

print(student)
```

## 13. `clear()`

Removes all items from the dictionary.

```python
student = {
    "name": "Shabbir",
    "age": 15
}

student.clear()

print(student)
```

## 14. `keys()`

Returns a view containing all dictionary keys.

```python
student = {
    "name": "Shabbir",
    "age": 15,
    "course": "Python"
}

print(student.keys())
```

## 15. `values()`

Returns a view containing all dictionary values.

```python
student = {
    "name": "Shabbir",
    "age": 15,
    "course": "Python"
}

print(student.values())
```

## 16. `items()`

Returns the dictionary's key-value pairs.

```python
student = {
    "name": "Shabbir",
    "age": 15,
    "course": "Python"
}

print(student.items())
```

## 17. `copy()`

Creates a shallow copy of a dictionary.

```python
student = {
    "name": "Shabbir",
    "age": 15
}

new_student = student.copy()

print("Original:", student)
print("Copy:", new_student)
```

## 18. `setdefault()`

Returns the value of a key. If the key does not exist, it creates the key with the given default value.

```python
student = {
    "name": "Shabbir",
    "age": 15
}

student.setdefault("course", "Python")
student.setdefault("age", 20)

print(student)
```

## 19. Membership Operators

Use `in` and `not in` to check whether a **key** exists in a dictionary.

```python
student = {
    "name": "Shabbir",
    "age": 15
}

print("name" in student)
print("email" not in student)
```

## 20. Loop Through Dictionary Keys

A normal `for` loop iterates through dictionary keys.

```python
student = {
    "name": "Shabbir",
    "age": 15,
    "course": "Python"
}

for key in student:
    print(key)
```

## 21. Loop Through Dictionary Values

Use `values()` to loop through values.

```python
student = {
    "name": "Shabbir",
    "age": 15,
    "course": "Python"
}

for value in student.values():
    print(value)
```

## 22. Loop Through Key-Value Pairs

Use `items()` to get both the key and value.

```python
student = {
    "name": "Shabbir",
    "age": 15,
    "course": "Python"
}

for key, value in student.items():
    print(key, ":", value)
```

## 23. Nested Dictionaries

A dictionary can contain another dictionary as a value.

```python
students = {
    "student1": {
        "name": "Shabbir",
        "marks": 85
    },
    "student2": {
        "name": "Ali",
        "marks": 90
    }
}

print(students)
print(students["student1"]["name"])
print(students["student2"]["marks"])
```

## 24. Dictionary Length

`len()` returns the number of key-value pairs.

```python
student = {
    "name": "Shabbir",
    "age": 15,
    "course": "Python"
}

print(len(student))
```

## 25. Creating a Dictionary with `dict()`

The `dict()` constructor can be used to create a dictionary.

```python
student = dict(name="Shabbir", age=15, course="Python")

print(student)
```

## 26. Dictionary Comprehension

Dictionary comprehension provides a short way to create dictionaries.

```python
squares = {number: number ** 2 for number in range(1, 6)}

print(squares)
```

## 27. `fromkeys()`

`fromkeys()` creates a new dictionary using the given keys and a common default value.

```python
keys = ["name", "age", "course"]

student = dict.fromkeys(keys, "Unknown")

print(student)
```

## 28. Useful Built-in Functions

These functions can be useful when working with dictionary keys, values, or items.

```python
student = {
    "name": "Shabbir",
    "age": 15,
    "marks": 85
}

print("Length:", len(student))
print("Keys:", list(student.keys()))
print("Values:", list(student.values()))
```

## 🧠 Dictionary Methods — Quick Revision

| Method | Use |
|---|---|
| `get()` | Safely access a value |
| `keys()` | Get all keys |
| `values()` | Get all values |
| `items()` | Get key-value pairs |
| `update()` | Add or update items |
| `pop()` | Remove a specific key |
| `popitem()` | Remove the last inserted pair |
| `clear()` | Remove all items |
| `copy()` | Create a copy |
| `setdefault()` | Get a value or create a default |
| `fromkeys()` | Create a dictionary from keys |

### Other useful operations

- `len()` — number of key-value pairs
- `in` — checks whether a key exists
- `not in` — checks whether a key does not exist
- `dict()` — creates a dictionary


## 🆚 List vs Tuple vs Dictionary

| Feature | List | Tuple | Dictionary |
|---|---|---|---|
| Syntax | `[]` | `()` | `{}` |
| Ordered | ✅ | ✅ | ✅ |
| Mutable | ✅ | ❌ | ✅ |
| Access | Index | Index | Key |
| Duplicate Values | ✅ | ✅ | ✅ |
| Key-Value Pairs | ❌ | ❌ | ✅ |

**Main idea:** A dictionary stores data using **keys and values**, which makes it useful when each piece of data needs a meaningful label.
